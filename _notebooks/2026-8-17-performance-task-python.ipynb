This notebook contains multiple code runner examples for a watch-themed CPT project.

Each example demonstrates one or more AP CSP concepts required for the Create Performance Task.

```python
# CODE_RUNNER: Output - display the watch catalog

watch_catalog = [
    'case', 'dial', 'hands', 'crown', 'movement', 'strap'
]

print('Watch catalog:')
for item in watch_catalog:
    print('-', item)

print('\nA watch has multiple important parts.')
```

```python
# CODE_RUNNER: Input - ask the user which watch part they want to learn about

watch_parts = ['case', 'dial', 'hands', 'crown', 'movement', 'strap']
info = {
    'case': 'Protects the movement and holds the watch together.',
    'dial': 'Shows the time and often includes markers or numbers.',
    'hands': 'Point to the current hour, minute, and second.',
    'crown': 'Used to set the time and wind a mechanical watch.',
    'movement': 'The internal mechanism that keeps time.',
    'strap': 'The band that holds the watch on the wrist.'
}

part = input('Enter a watch part to learn about: ').strip().lower()

if part == '':
    print('You entered nothing. Here are the watch parts:')
    for item in watch_parts:
        print('-', item)
elif part in info:
    print(part.capitalize() + ':', info[part])
else:
    print('Unknown part. Try:', ', '.join(watch_parts))
```

```python
# CODE_RUNNER: List - store multiple watch products in a list

watch_inventory = [
    'AeroLite 42',
    'Crest 36',
    'Nova X',
    'Heritage M1'
]

print('Current inventory:')
for watch in watch_inventory:
    print('-', watch)

print('\nTotal watches:', len(watch_inventory))
```

```python
# CODE_RUNNER: Procedure - a reusable function to find a watch by name

def find_watch(watches, target):
    for index in range(len(watches)):
        if watches[index].lower() == target.lower():
            return index
    return -1

inventory = ['AeroLite 42', 'Crest 36', 'Nova X', 'Heritage M1']
search = input('Enter a watch name to search for: ').strip()
position = find_watch(inventory, search)

if position == -1:
    print(search, 'was not found in the inventory.')
else:
    print(search, 'was found at index', position)
```

```python
# CODE_RUNNER: Sequence - do tasks in order to calculate a watch total

watch_price = 120
tax_rate = 0.08
shipping = 12
subtotal = watch_price * 1
tax = subtotal * tax_rate
total = subtotal + tax + shipping

print('Watch price:', watch_price)
print('Tax:', tax)
print('Shipping:', shipping)
print('Final total:', total)
```

```python
# CODE_RUNNER: Selection - decide a watch category by price

price = int(input('Enter the watch price: '))

if price < 100:
    category = 'Budget'
elif price < 200:
    category = 'Mid-range'
else:
    category = 'Premium'

print('Watch category:', category)
```

```python
# CODE_RUNNER: Iteration - loop through each watch in the inventory

inventory = [
    {'name': 'AeroLite 42', 'price': 120, 'stock': 8},
    {'name': 'Crest 36', 'price': 90, 'stock': 0},
    {'name': 'Nova X', 'price': 220, 'stock': 5},
    {'name': 'Heritage M1', 'price': 310, 'stock': 3}
]

for watch in inventory:
    if watch['stock'] > 0:
        status = 'available'
    else:
        status = 'out of stock'
    print(watch['name'], '-', '$' + str(watch['price']), '-', status)
```

```python
# CODE_RUNNER: Algorithm - combine input, list, selection, and iteration

inventory = [
    {'name': 'AeroLite 42', 'brand': 'Timex', 'price': 120, 'stock': 8},
    {'name': 'Crest 36', 'brand': 'Casio', 'price': 90, 'stock': 0},
    {'name': 'Nova X', 'brand': 'Citizen', 'price': 220, 'stock': 5},
    {'name': 'Heritage M1', 'brand': 'Seiko', 'price': 310, 'stock': 3}
]

budget = int(input('What is your maximum budget? '))
recommended = []

for watch in inventory:
    if watch['stock'] > 0 and watch['price'] <= budget:
        recommended.append(watch['name'])

if len(recommended) == 0:
    print('No watches fit your budget right now.')
else:
    print('Recommended watches:')
    for item in recommended:
        print('-', item)
```

```python
# CODE_RUNNER: List operations - append, insert, remove, and length

watch_list = ['AeroLite 42', 'Crest 36']
print('Start:', watch_list)

watch_list.append('Nova X')
print('After append:', watch_list)

watch_list.insert(1, 'Classic 7')
print('After insert:', watch_list)

watch_list.remove('Crest 36')
print('After remove:', watch_list)

print('Length:', len(watch_list))
```

```python
# CODE_RUNNER: Search algorithm - linear search through watch brands

brands = ['Timex', 'Casio', 'Citizen', 'Seiko', 'Omega']
search_brand = input('Enter a brand to search for: ').strip()

found = False
for index in range(len(brands)):
    if brands[index].lower() == search_brand.lower():
        print(search_brand, 'found at index', index)
        found = True
        break

if not found:
    print(search_brand, 'was not found in the list.')
```

```python
# CODE_RUNNER: Boolean logic - decide if a watch is a good purchase

watch = {
    'name': 'Nova X',
    'price': 220,
    'stock': 5,
    'brand': 'Citizen'
}

budget = int(input('Enter your budget: '))

is_affordable = watch['price'] <= budget
has_stock = watch['stock'] > 0
is_good_brand = watch['brand'] == 'Citizen' or watch['brand'] == 'Seiko'

if is_affordable and has_stock and is_good_brand:
    print('This watch is a strong purchase option.')
else:
    print('This watch may not be the best choice right now.')
```
