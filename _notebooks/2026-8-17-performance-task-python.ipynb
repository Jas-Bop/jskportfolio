## Project Description: Watch Inventory Manager

For our project, we are making a watch inventory manager. We chose this idea because watches have multiple parts and variations, and we wanted a program that could help us track them. Our program keeps track of watch models, brands, prices, and stock levels. It can display inventory, search for specific watches, filter by price range, and recommend watches based on budget.

This notebook follows the CPT concept flow from the Open Coding lesson and changes the project theme to watches. Each Python code runner shows one part of a complete program: input, output, lists, procedures, sequencing, selection, iteration, algorithms, list operations, search, and boolean logic.

## Output

Output is information shown to the user. In Python, output uses `print()` statements.


```python
# CODE_RUNNER: Output a watch and its details

watch_name = "Nova X"
brand = "Citizen"
price = 220

print("Watch: " + watch_name)
print("Brand: " + brand)
print("Price: $" + str(price))
```

```python
# CODE_RUNNER: Use boolean logic to decide if watch is a good purchase

watch = {'name': 'Nova X', 'price': 220, 'stock': 5, 'brand': 'Citizen'}
budget = int(input('Your budget: '))

is_affordable = watch['price'] <= budget
has_stock = watch['stock'] > 0
is_quality_brand = watch['brand'] == 'Citizen' or watch['brand'] == 'Seiko'

if is_affordable and has_stock and is_quality_brand:
    print('Recommendation: Great purchase!')
else:
    print('Recommendation: Consider other options.')
```

## Boolean Logic

Boolean logic uses AND, OR, and NOT to combine conditions for complex decisions.


```python
# CODE_RUNNER: Implement linear search through watch brands

brands = ['Timex', 'Casio', 'Citizen', 'Seiko', 'Omega']
search_brand = input('Enter a brand to search: ').strip()

found = False
for index in range(len(brands)):
    if brands[index].lower() == search_brand.lower():
        print('Found at index:', index)
        found = True
        break

if not found:
    print('Brand not found.')
```

## Search Algorithm

A search algorithm finds an item in a list. Linear search checks each item until found.


```python
# CODE_RUNNER: Demonstrate append, insert, remove, and length

inventory = ['AeroLite 42', 'Crest 36']
print('Start:', inventory)

inventory.append('Nova X')
print('After append:', inventory)

inventory.insert(1, 'Classic 7')
print('After insert at index 1:', inventory)

inventory.remove('Crest 36')
print('After remove:', inventory)

print('Length:', len(inventory))
```

## List Operations

List operations include append (add), insert (add at index), remove (delete), and length (count).


```python
# CODE_RUNNER: Find watches within budget (complete algorithm)

inventory = [
    {'name': 'AeroLite 42', 'price': 120, 'stock': 8},
    {'name': 'Crest 36', 'price': 90, 'stock': 0},
    {'name': 'Nova X', 'price': 220, 'stock': 5},
    {'name': 'Heritage M1', 'price': 310, 'stock': 3}
]

budget = int(input('What is your maximum budget? '))
recommended = []

for watch in inventory:
    if watch['stock'] > 0 and watch['price'] <= budget:
        recommended.append(watch['name'])

if len(recommended) == 0:
    print('No watches fit your budget.')
else:
    print('Watches within budget:')
    for name in recommended:
        print('-', name)
```

## Algorithm

A complete algorithm combines sequencing, selection, and iteration to solve a meaningful problem.


```python
# CODE_RUNNER: Loop through watch inventory and show status

inventory = [
    {'name': 'AeroLite 42', 'price': 120, 'stock': 8},
    {'name': 'Crest 36', 'price': 90, 'stock': 0},
    {'name': 'Nova X', 'price': 220, 'stock': 5}
]

for watch in inventory:
    status = 'Available' if watch['stock'] > 0 else 'Out of stock'
    print(watch['name'] + ' - $' + str(watch['price']) + ' - ' + status)
```

## Iteration

Iteration repeats code. Loops process every item in a list without duplicate code.


```python
# CODE_RUNNER: Decide if a watch is in stock and recommend it

stock = int(input('How many in stock? '))
price = int(input('Price? '))

if stock > 0 and price < 200:
    print('Recommendation: Great watch to buy!')
elif stock > 0:
    print('Recommendation: In stock but pricey.')
else:
    print('Out of stock.')
```

## Selection

Selection uses conditions so a program can choose different paths based on IF/ELSE.


```python
# CODE_RUNNER: Calculate watch total with tax and shipping

watch_price = 120
tax_rate = 0.08
shipping = 12

subtotal = watch_price
tax = subtotal * tax_rate
total = subtotal + tax + shipping

print('Price: $' + str(watch_price))
print('Tax: $' + str(tax))
print('Shipping: $' + str(shipping))
print('Total: $' + str(total))
```

## Sequence

Sequence means statements run in order. Calculations depend on earlier values.


```python
# CODE_RUNNER: Create a function to classify watch price range

def get_price_category(price):
    if price < 100:
        return 'Budget'
    elif price < 200:
        return 'Mid-range'
    else:
        return 'Premium'

watch_price = int(input('Enter watch price: '))
category = get_price_category(watch_price)
print('Category:', category)
```

## Procedure

A procedure is a named block of reusable code. In Python, procedures are defined with `def`.


```python
# CODE_RUNNER: Store watch models in a list and display the count

watch_models = ['AeroLite 42', 'Crest 36', 'Nova X', 'Heritage M1']

print('Watch inventory:')
for model in watch_models:
    print('-', model)

print('Total models:', len(watch_models))
```

## List

Lists store multiple values. Python lists use zero-based indexing.


```python
# CODE_RUNNER: Ask the user for a watch brand to search

watch_brands = ['Timex', 'Casio', 'Citizen', 'Seiko', 'Omega']
search_brand = input('Enter a watch brand: ').strip()

if search_brand in watch_brands:
    print('Found: ' + search_brand)
else:
    print('Brand not found. Available:', ', '.join(watch_brands))
```
