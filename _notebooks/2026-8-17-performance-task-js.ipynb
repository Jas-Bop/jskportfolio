## List

Lists store multiple values. JavaScript uses arrays, and array indexes start at `0`.

```javascript
%%js

// CODE_RUNNER: Store watch parts in an array and display the inventory count.

const watchParts = ["case", "crown", "movement", "strap"];

console.log("Parts list:");
console.log(watchParts);
console.log("Total parts tracked: " + watchParts.length);
```

## Procedure

A procedure is a named block of reusable code. In JavaScript, procedures are usually written as functions.

```javascript
%%js

// CODE_RUNNER: Create and call a function that rates a watch part condition.

function conditionLabel(score) {
  if (score >= 90) {
    return "excellent";
  } else if (score >= 70) {
    return "usable";
  } else {
    return "needs repair";
  }
}

console.log("Crystal condition: " + conditionLabel(82));
```

## Sequence

Sequence means statements run in order. This matters when calculations depend on earlier values.

```javascript
%%js

// CODE_RUNNER: Calculate a watch repair estimate step by step.

const crownCost = 18;
const strapCost = 35;
const laborCost = 42;
const subtotal = crownCost + strapCost + laborCost;
const tax = subtotal * 0.08;
const total = subtotal + tax;

console.log("Subtotal: $" + subtotal.toFixed(2));
console.log("Tax: $" + tax.toFixed(2));
console.log("Total estimate: $" + total.toFixed(2));
```

## Selection

Selection uses conditions so a program can choose different paths.

```javascript
%%js

// CODE_RUNNER: Decide whether a watch needs a full service based on age and accuracy.

const yearsSinceService = 6;
const secondsOffPerDay = 24;

if (yearsSinceService >= 5 || secondsOffPerDay > 20) {
  console.log("Recommendation: full service needed.");
} else {
  console.log("Recommendation: routine inspection is enough.");
}
```

## Iteration

Iteration repeats steps. Loops help process every item in a list without duplicate code.

```javascript
%%js

// CODE_RUNNER: Loop through watch parts and print an inspection checklist.

const inspectionParts = ["case", "dial", "hands", "crown", "movement", "strap"];

for (const part of inspectionParts) {
  console.log("Inspect: " + part);
}
```

## Algorithm

A complete CPT algorithm combines sequencing, selection, and iteration to solve a meaningful problem.

```javascript
%%js

// CODE_RUNNER: Find the average condition score and recommend a repair plan.

const partScores = [95, 88, 61, 74, 92, 68];
let totalScore = 0;
let weakPartCount = 0;

for (const score of partScores) {
  totalScore += score;
  if (score < 70) {
    weakPartCount += 1;
  }
}

const averageScore = totalScore / partScores.length;
console.log("Average condition: " + averageScore.toFixed(1));

if (weakPartCount > 0) {
  console.log("Repair plan: replace or repair " + weakPartCount + " weak part(s).");
} else {
  console.log("Repair plan: clean and polish only.");
}
```

## List Operations

List operations add, remove, update, and count values. JavaScript arrays use methods such as `push()`, `splice()`, and `includes()`.

```javascript
%%js

// CODE_RUNNER: Add, insert, remove, and count watch parts in an inventory array.

const inventory = ["case", "dial", "hands"];
inventory.push("crown");
inventory.splice(1, 0, "crystal");
inventory.splice(inventory.indexOf("hands"), 1);

console.log("Updated inventory:");
console.log(inventory);
console.log("Inventory count: " + inventory.length);
```

## Search Algorithm

Linear search checks list items one at a time until it finds a target or reaches the end.

```javascript
%%js

// CODE_RUNNER: Search for a watch part by name using a linear search.

function findPart(parts, target) {
  for (let index = 0; index < parts.length; index++) {
    if (parts[index] === target) {
      return index;
    }
  }
  return -1;
}

const partsCatalog = ["case", "dial", "hands", "crown", "movement", "strap"];
const targetPart = "movement";
const foundIndex = findPart(partsCatalog, targetPart);

if (foundIndex >= 0) {
  console.log(targetPart + " found at JavaScript index " + foundIndex + ".");
} else {
  console.log(targetPart + " was not found.");
}
```

## Boolean Logic

Boolean logic combines true/false conditions with `&&`, `||`, and `!`.

```javascript
%%js

// CODE_RUNNER: Use boolean logic to decide if a watch part can be reused.

const part = "strap";
const conditionScore = 76;
const isOriginal = true;
const isSafetyCritical = false;

const canReuse = conditionScore >= 75 && (isOriginal || !isSafetyCritical);

console.log("Part: " + part);
console.log("Can reuse? " + canReuse);
```
