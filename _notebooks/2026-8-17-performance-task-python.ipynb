## Output

Output is information shown to the user. In Python, output uses `print()` statements.


```python
# CODE_RUNNER: Output a watch and its details | height: 560px

watch_name = "Nova X"
brand = "Citizen"
price = 220

print("Watch: " + watch_name)
print("Brand: " + brand)
print("Price: $" + str(price))
```

```python
# CODE_RUNNER: Use boolean logic to decide if watch is a good purchase | height: 560px

watch = {'name': 'Nova X', 'price': 220, 'stock': 5, 'brand': 'Citizen'}
budget = int(input('Your budget: '))

is_affordable = watch['price'] <= budget
has_stock = watch['stock'] > 0
is_quality_brand = watch['brand'] == 'Citizen' or watch['brand'] == 'Seiko'

if is_affordable and has_stock and is_quality_brand:
    print('Recommendation: Great purchase!')
else:
    print('Recommendation: Consider other options.')
```

## Boolean Logic

Boolean logic uses AND, OR, and NOT to combine conditions for complex decisions.


```python
# CODE_RUNNER: Implement search through a watch brand dictionary | height: 560px

brands = {
    'Timex': 'United States',
    'Casio': 'Japan',
    'Citizen': 'Japan',
    'Seiko': 'Japan',
    'Omega': 'Switzerland'
}
search_brand = input('Enter a brand to search: ').strip()

found = False
for brand, country in brands.items():
    if brand.lower() == search_brand.lower():
        print('Found:', brand)
        print('Country:', country)
        found = True
        break

if not found:
    print('Brand not found.')
```

## Search Algorithm

A search algorithm finds a target in a data structure. With dictionaries, a program can search keys and then use the matching value.


```python
# CODE_RUNNER: Demonstrate add, update, remove, and length with a dictionary | height: 560px

inventory = {
    'AeroLite 42': 120,
    'Crest 36': 90
}
print('Start:', inventory)

inventory['Nova X'] = 220
print('After adding Nova X:', inventory)

inventory['AeroLite 42'] = 135
print('After updating AeroLite 42 price:', inventory)

del inventory['Crest 36']
print('After removing Crest 36:', inventory)

print('Length:', len(inventory))
```

## Dictionary Operations

Dictionary operations include adding a key, updating a value, deleting a key, and checking length.


```python
# CODE_RUNNER: Find watches within budget (complete algorithm) | height: 560px

inventory = {
    'AeroLite 42': {'price': 120, 'stock': 8},
    'Crest 36': {'price': 90, 'stock': 0},
    'Nova X': {'price': 220, 'stock': 5},
    'Heritage M1': {'price': 310, 'stock': 3}
}

budget = int(input('What is your maximum budget? '))
recommended = {}

for model, details in inventory.items():
    if details['stock'] > 0 and details['price'] <= budget:
        recommended[model] = details['price']

if len(recommended) == 0:
    print('No watches fit your budget.')
else:
    print('Watches within budget:')
    for model, price in recommended.items():
        print('-', model, '$' + str(price))
```

## Algorithm

A complete algorithm combines sequencing, selection, and iteration to solve a meaningful problem.


```python
# CODE_RUNNER: Loop through watch inventory and show status | height: 560px

inventory = {
    'AeroLite 42': {'price': 120, 'stock': 8},
    'Crest 36': {'price': 90, 'stock': 0},
    'Nova X': {'price': 220, 'stock': 5}
}

for model, details in inventory.items():
    status = 'Available' if details['stock'] > 0 else 'Out of stock'
    print(model + ' - $' + str(details['price']) + ' - ' + status)
```

## Iteration

Iteration repeats code. Loops process every key-value pair in a dictionary without duplicate code.


```python
# CODE_RUNNER: Decide if a watch is in stock and recommend it | height: 560px

stock = int(input('How many in stock? '))
price = int(input('Price? '))

if stock > 0 and price < 200:
    print('Recommendation: Great watch to buy!')
elif stock > 0:
    print('Recommendation: In stock but pricey.')
else:
    print('Out of stock.')
```

## Selection

Selection uses conditions so a program can choose different paths based on IF/ELSE.


```python
# CODE_RUNNER: Calculate watch total with tax and shipping | height: 560px

watch_price = 120
tax_rate = 0.08
shipping = 12

subtotal = watch_price
tax = subtotal * tax_rate
total = subtotal + tax + shipping

print('Price: $' + str(watch_price))
print('Tax: $' + str(tax))
print('Shipping: $' + str(shipping))
print('Total: $' + str(total))
```

## Sequence

Sequence means statements run in order. Calculations depend on earlier values.


```python
# CODE_RUNNER: Create a function to classify watch price range | height: 560px

def get_price_category(price):
    if price < 100:
        return 'Budget'
    elif price < 200:
        return 'Mid-range'
    else:
        return 'Premium'

watch_price = int(input('Enter watch price: '))
category = get_price_category(watch_price)
print('Category:', category)
```

## Procedure

A procedure is a named block of reusable code. In Python, procedures are defined with `def`.


```python
# CODE_RUNNER: Store watch models in a dictionary and display the count | height: 560px

watch_models = {
    'AeroLite 42': 'sport',
    'Crest 36': 'dress',
    'Nova X': 'field',
    'Heritage M1': 'classic'
}

print('Watch inventory:')
for model, style in watch_models.items():
    print('-', model, '(' + style + ')')

print('Total models:', len(watch_models))
```

## Dictionary

Dictionaries store key-value pairs. Python dictionaries let a program look up details by a meaningful key.


```python
# CODE_RUNNER: Ask the user for a watch brand to search | height: 560px

watch_brands = {
    'Timex': 'United States',
    'Casio': 'Japan',
    'Citizen': 'Japan',
    'Seiko': 'Japan',
    'Omega': 'Switzerland'
}
search_brand = input('Enter a watch brand: ').strip()

if search_brand in watch_brands:
    print('Found: ' + search_brand)
    print('Country: ' + watch_brands[search_brand])
else:
    print('Brand not found. Available:', ', '.join(watch_brands.keys()))
```
