## Dictionary

Dictionaries store key-value pairs. In JavaScript, plain objects can work like dictionaries.

```javascript
%%js

// CODE_RUNNER: Store watch parts in an object dictionary and display the inventory count.

const watchParts = {
  case: "protects the movement",
  crown: "sets the time",
  movement: "powers the watch",
  strap: "holds the watch on the wrist"
};

console.log("Parts dictionary:");
for (const part in watchParts) {
  console.log(part + ": " + watchParts[part]);
}
console.log("Total parts tracked: " + Object.keys(watchParts).length);
```

## Procedure

A procedure is a named block of reusable code. In JavaScript, procedures are usually written as functions.

```javascript
%%js

// CODE_RUNNER: Create and call a function that rates a watch part condition.

function conditionLabel(score) {
  if (score >= 90) {
    return "excellent";
  } else if (score >= 70) {
    return "usable";
  } else {
    return "needs repair";
  }
}

console.log("Crystal condition: " + conditionLabel(82));
```

## Sequence

Sequence means statements run in order. This matters when calculations depend on earlier values.

```javascript
%%js

// CODE_RUNNER: Calculate a watch repair estimate step by step.

const crownCost = 18;
const strapCost = 35;
const laborCost = 42;
const subtotal = crownCost + strapCost + laborCost;
const tax = subtotal * 0.08;
const total = subtotal + tax;

console.log("Subtotal: $" + subtotal.toFixed(2));
console.log("Tax: $" + tax.toFixed(2));
console.log("Total estimate: $" + total.toFixed(2));
```

## Selection

Selection uses conditions so a program can choose different paths.

```javascript
%%js

// CODE_RUNNER: Decide whether a watch needs a full service based on age and accuracy.

const yearsSinceService = 6;
const secondsOffPerDay = 24;

if (yearsSinceService >= 5 || secondsOffPerDay > 20) {
  console.log("Recommendation: full service needed.");
} else {
  console.log("Recommendation: routine inspection is enough.");
}
```

## Iteration

Iteration repeats steps. Loops help process every key-value pair in a dictionary without duplicate code.

```javascript
%%js

// CODE_RUNNER: Loop through a watch parts dictionary and print an inspection checklist.

const inspectionParts = {
  case: "check scratches",
  dial: "check stains",
  hands: "check alignment",
  crown: "check winding",
  movement: "check accuracy",
  strap: "check wear"
};

for (const part in inspectionParts) {
  console.log("Inspect " + part + ": " + inspectionParts[part]);
}
```

## Algorithm

A complete CPT algorithm combines sequencing, selection, and iteration to solve a meaningful problem.

```javascript
%%js

// CODE_RUNNER: Find the average condition score and recommend a repair plan.

const partScores = {
  case: 95,
  dial: 88,
  hands: 61,
  crown: 74,
  movement: 92,
  strap: 68
};
let totalScore = 0;
let weakPartCount = 0;
let partCount = 0;

for (const part in partScores) {
  const score = partScores[part];
  totalScore += score;
  partCount += 1;
  if (score < 70) {
    weakPartCount += 1;
  }
}

const averageScore = totalScore / partCount;
console.log("Average condition: " + averageScore.toFixed(1));

if (weakPartCount > 0) {
  console.log("Repair plan: replace or repair " + weakPartCount + " weak part(s).");
} else {
  console.log("Repair plan: clean and polish only.");
}
```

## Dictionary Operations

Dictionary operations add, remove, update, and count key-value pairs. JavaScript objects can use bracket notation and `delete`.

```javascript
%%js

// CODE_RUNNER: Add, update, remove, and count watch parts in an inventory object.

const inventory = {
  case: "good",
  dial: "fair",
  hands: "replace",
  movement: "good",
  strap: "worn",
  buckle: "good"
};
inventory.crown = "inspect";
inventory.crystal = "polish";
inventory.dial = "clean";
delete inventory.hands;

console.log("Updated inventory:");
console.log(inventory);
console.log("Inventory count: " + Object.keys(inventory).length);
```

## Search Algorithm

Search checks dictionary keys one at a time until it finds a target or reaches the end.

```javascript
%%js

// CODE_RUNNER: Search for a watch part by name using an object dictionary.

function findPart(parts, target) {
  for (const part in parts) {
    if (part === target) {
      return parts[part];
    }
  }
  return null;
}

const partsCatalog = {
  case: "outer shell",
  dial: "display face",
  hands: "time indicators",
  crown: "time setting control",
  movement: "engine of the watch",
  strap: "wrist attachment"
};
const targetPart = "movement";
const foundDescription = findPart(partsCatalog, targetPart);

if (foundDescription !== null) {
  console.log(targetPart + " found: " + foundDescription + ".");
} else {
  console.log(targetPart + " was not found.");
}
```

## Boolean Logic

Boolean logic combines true/false conditions with `&&`, `||`, and `!`.

```javascript
%%js

// CODE_RUNNER: Use boolean logic to decide if a watch part can be reused.

const part = "strap";
const conditionScore = 76;
const isOriginal = true;
const isSafetyCritical = false;

const canReuse = conditionScore >= 75 && (isOriginal || !isSafetyCritical);

console.log("Part: " + part);
console.log("Can reuse? " + canReuse);
```
