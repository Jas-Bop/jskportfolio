Our program keeps track of parts like the case, dial, hands, crown, movement, crystal, and strap. This lesson uses our watch parts tracker to show what College Board expects: algorithms that solve a real problem, procedures that accept parameters and return values, lists that store and process data, and boolean logic that makes decisions. When we see how these concepts connect in a themed project, we can design our own CPT program that earns credit for algorithm implementation, abstraction, and program purpose.

## List

**CPT Requirement:** Our program must use a list to manage multiple data values. In JavaScript, lists are arrays. Lists help us store and process multiple watch parts without making a separate variable for every part.

### Code Runner Challenge
Store and total watch part condition scores.

```javascript
%%js

// CODE_RUNNER: Store and total watch part condition scores.

let partScores = [95, 88, 76, 91];
let total = 0;

for (let score of partScores) {
  total = total + score;
}

console.log("Condition Scores: " + partScores);
console.log("Total Condition Points: " + total);
```

## Procedure

**CPT Requirement:** We must create at least one student-developed procedure with parameters. In JavaScript, we use functions. This demonstrates abstraction because we can reuse a task.

### Code Runner Challenge
Determine a watch part condition label from a score.

```javascript
%%js

// CODE_RUNNER: Determine a watch part condition label from a score.

function getConditionLabel(score) {
  if (score >= 90) {
    return "excellent";
  } else if (score >= 75) {
    return "good";
  } else {
    return "needs repair";
  }
}

let label = getConditionLabel(88);
console.log("Crown condition: " + label);
```

## Sequence

**Exam Concept:** Sequencing means statements execute in order, line by line. In our watch parts tracker, sequence matters because we calculate costs step by step before displaying the final repair estimate.

### Code Runner Challenge
Calculate a watch repair estimate step by step.

```javascript
%%js

// CODE_RUNNER: Calculate a watch repair estimate step by step.

let crystalCost = 24;
let strapCost = 35;
let laborCost = 40;
let subtotal = crystalCost + strapCost + laborCost;
let tax = subtotal * 0.08;
let totalCost = subtotal + tax;

console.log("Subtotal: $" + subtotal.toFixed(2));
console.log("Tax: $" + tax.toFixed(2));
console.log("Final Estimate: $" + totalCost.toFixed(2));
```

## Selection

**CPT Requirement:** Our algorithm must include selection using `if` or `if/else` statements. Selection helps decide if a part should be reused, repaired, or replaced.

### Code Runner Challenge
Make a repair decision using selection.

```javascript
%%js

// CODE_RUNNER: Make a repair decision using selection.

let score = Number(prompt("Part condition score: 0 - 100") || 82);
let action = "Unknown";

if (score >= 90) {
  action = "reuse the part";
} else if (score >= 70) {
  action = "repair the part";
} else {
  action = "replace the part";
}

console.log("Score: " + score);
console.log("Recommended Action: " + action);
```

## Iteration

**CPT Requirement:** Our algorithm must include iteration, which means loops. Iteration lets our program process multiple watch parts without duplicate code.

### Code Runner Challenge
Enter watch parts until done using a sentinel pattern.

```javascript
%%js

// CODE_RUNNER: Enter watch parts until done using a sentinel pattern.

let parts = [];
let continueEntry = "yes";

while (continueEntry !== "no") {
  let part = prompt("Add a watch part:") || "case";
  parts.push(part);
  continueEntry = (prompt("Continue entering parts? Type no to stop") || "no").toLowerCase();
}

console.log("List of parts: " + parts);
console.log("Total parts entered: " + parts.length);
```

## Algorithm

**CPT Requirement:** Our algorithm must integrate sequencing, selection, and iteration to solve a meaningful problem. In our watch parts tracker, the algorithm can calculate average condition and decide an overall repair plan.

### Code Runner Challenge
Find average condition from a list and recommend a plan.

```javascript
%%js

// CODE_RUNNER: Find average condition from a list and recommend a plan.

function calculateAverage(numbers) {
  let total = 0;
  for (let num of numbers) {
    total = total + num;
  }

  if (numbers.length > 0) {
    return total / numbers.length;
  } else {
    return 0;
  }
}

let scores = [92, 80, 68, 95, 74];
let average = calculateAverage(scores);

if (average >= 85) {
  console.log("Overall Plan: light cleaning");
} else {
  console.log("Overall Plan: full repair check");
}

console.log("Average Condition: " + average.toFixed(1));
```

## Search Algorithm

**Exam Topic:** Linear search uses iteration through a list, selection to check conditions, and a return value. In our project, search helps us find a part in the watch inventory.

### Code Runner Challenge
Search for a watch part in the inventory.

```javascript
%%js

// CODE_RUNNER: Search for a watch part in the inventory.

function findPart(partList, targetPart) {
  for (let index = 0; index < partList.length; index++) {
    if (partList[index] === targetPart) {
      return index;
    }
  }
  return -1;
}

let inventory = ["case", "dial", "hands", "crown", "movement", "strap"];
let searchFor = (prompt("Enter part you are looking for:") || "movement").toLowerCase();
let position = findPart(inventory, searchFor);

if (position >= 0) {
  console.log(searchFor + " found at JavaScript index " + position);
} else {
  console.log(searchFor + " was not found");
}
```

## Boolean Logic

**Exam Concept:** JavaScript uses `&&`, `||`, and `!` to combine conditions. Boolean expressions evaluate to true or false. In our watch parts tracker, boolean logic helps us decide if a part is safe to reuse.

### Code Runner Challenge
Use boolean logic to decide if a watch part can be reused.

```javascript
%%js

// CODE_RUNNER: Use boolean logic to decide if a watch part can be reused.

let part = "crystal";
let condition = 82;
let isOriginal = true;
let isSafetyCritical = true;
let canReuse = condition >= 80 && (isOriginal || !isSafetyCritical);

console.log("Part: " + part);
console.log("Can reuse? " + canReuse);
```
