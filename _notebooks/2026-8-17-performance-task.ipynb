```python
# CODE_RUNNER: Watch parts and brief history

# Output: list parts and simple descriptions
watch_parts = ['case', 'dial', 'hands', 'crown', 'movement', 'strap']
history = {
    'case': 'Protects the movement; historically metal or precious metals.',
    'dial': 'Shows the time; designs evolved over centuries.',
    'hands': 'Point to hours/minutes/seconds; shapes vary.',
    'crown': 'Used to wind or set time; developed with mechanical watches.',
    'movement': 'The engine (mechanical or quartz) that keeps time.',
    'strap': 'Attaches watch to wrist; materials changed with fashion.'
}
print('Watch parts:', watch_parts)
print()
part = input('Enter a watch part to learn about (or press enter to list all): ').strip().lower()
if part == '':
    print('List of parts:')
    for p in watch_parts:
        print('-', p)
elif part in history:
    print(part.capitalize() + ':', history[part])
else:
    print('Unknown part. Try one of:', ', '.join(watch_parts))

# Procedure example
def count_parts(parts):
    return len(parts)

print('\nTotal parts tracked:', count_parts(watch_parts))

# Iteration example: show lengths
for p in watch_parts:
    print(p + ' length:', len(p))
```
