# DSA Website Ideation

This section focuses on the DSA website and the problems that make it difficult for visitors to use.

## Problems with the Current Website

The current DSASD.org experience creates friction for both members and the public. Important information is spread across multiple pages, PDFs, and external links, which makes the site feel outdated and difficult to trust.

* **Information is scattered:** Benefits, board updates, forms, and event details are not grouped in a clear way.
* **Member and public content are mixed together:** People do not know whether they are looking at information for members or for the public.
* **Search and navigation are weak:** Visitors have to click through too many pages just to find a form or document.
* **Important pages feel buried:** Donation information, community events, and member resources are not easy to find from the main experience.
* **Static PDFs are harder to use:** Many documents are posted as PDFs instead of clear, readable web pages.
* **The homepage does not guide people:** There is no clear path for what a visitor should do next.
* **The user experience feels outdated:** The site does not create a strong sense of trust, clarity, or modern organization.
* **The login button says log out when you haven't even logged in yet.**

## How We Can Fix It

### 1. Separate the two audiences clearly
Create a visible top toggle or split navigation with two paths:

* **Public Hub:** events, donation links, community news, foundation updates
* **Member Portal:** benefits, legal resources, board docs, forms, retirement information

### 2. Build a cleaner dashboard for members
Create a single landing page with sections for:

* Benefits and enrollment
* Legal assistance resources
* Board updates and meeting minutes
* Discounts and partner programs
* Forms and documents

### 3. Simplify the public experience
Focus on trust and action with clear calls to action like:

* Donate
* View community events
* Read news updates
* Contact the foundation

### 4. Replace scattered PDFs with cleaner digital pages
Use readable web pages instead of burying information in PDFs.

### 5. Improve events and forms
Add search, filters, categories, RSVP links, and digital forms.

### 6. Improve search and homepage guidance
Use a visible search bar and clear buttons like “Member Login,” “Benefits,” “Donate,” and “Events.”

### 7. Make the site feel more organized and trustworthy
Use consistent page structure, cleaner hierarchy, and stronger visual contrast.

```mermaid
flowchart TD
    A[Visitor lands on homepage] --> B{What are they looking for?}
    B -->|Member resources| C[Member Portal]
    B -->|Public information| D[Public Hub]
    C --> E[Benefits, forms, legal docs]
    D --> F[Events, news, donations]
    E --> G[Action complete]
    F --> G
    G --> H[Trust restored]
```

## Summary

The main problem is confusion. Visitors cannot quickly tell where to go, what is for members, and what is for the public. The fix is to organize the site around user needs, group important information, reduce reliance on PDFs, and give each audience a clear path.

# RCR: Poway-Midland Railroad Digital Experience Ideation

This section focuses on the Poway-Midland Railroad project as a design-based research capstone. The goal was to transform a static informational website into a more useful, dynamic platform for visitors, staff, and volunteers.

## Project Vision

The original PMRR website was mostly a static list of dates and times. It gave visitors basic information, but it did not help them plan a trip, understand operations, or interact with the railroad in a meaningful way.

The capstone redesign focused on helping users answer real questions such as:

* When is the next train?
* Are there seats available?
* What is the best time to visit?
* Can I book a reservation online?
* What is the history of the railroad?
* How can I volunteer or support the park?

## Main Problems in the Original Experience

* **Static information only:** The original site showed dates and times but provided no live operational details.
* **No seat visibility:** Visitors could not check whether a ride was full before driving out.
* **No online booking:** Ticket reservations had to be handled in person, which created friction.
* **No visitor planning tools:** Guests had no way to predict crowd levels or plan around busy days.
* **History and fleet info was hard to explore:** The site had text-heavy sections that were difficult to browse.
* **No digital volunteer support:** Volunteer scheduling and shift sign-up were not available online.
* **No community engagement tools:** Visitors had no place to leave notes, comments, or photos.
* **No AI support:** Basic questions still required staff time and manual responses.

## What We Built

The redesigned PMRR experience introduced several features designed to improve both visitor usefulness and internal operations.

### 1. Live Train Schedule and Departure Board
A real-time departure board replaced the static table and allowed users to check seat availability and operating status.

### 2. Online Booking and Reservation System
Visitors could select a ride, choose ticket count, and receive a confirmation code. This reduced overbooking risk and improved planning.

### 3. Visitor Forecast with Machine Learning
The system used weather, holidays, school calendars, and schedule trend data to predict crowd size and help visitors plan smarter.

### 4. Interactive Fleet Roster
The fleet was redesigned as a visual, interactive archive with photos, filters, specs, and restoration details.

### 5. Interactive History Timeline
The railroad history was transformed from long text sections into a clickable, visualized timeline with major eras and important figures.

### 6. Volunteer Scheduling Portal
Volunteers could view future shifts, sign up, cancel, and track coverage with live scheduling data.

### 7. Event Search and Announcement Board
An updated layout let visitors search and filter events by category instead of scanning an unorganized list.

### 8. 360° Panorama and Interactive Route Map
This added virtual exploration for users before or after visiting the site and made the experience more immersive.

### 9. Community Notes Board
Visitors could post notes and photos, creating a sense of shared history and engagement.

### 10. AI Visitor Assistant
A floating assistant gave quick guidance for common visitor questions, reducing the time staff had to spend answering repetitive inquiries.

### 11. Membership and Donation Portal
Supporters could join, donate, and manage their membership in a clearer online flow.

---

## What the Redesign Solved

The PMRR project showed that a historic site can still benefit from modern digital tools. The redesign improved the experience by making it more useful, more engaging, and more responsive to real user needs.

* **Visitors could plan ahead better**
* **Staff could manage operations more efficiently**
* **The site felt more modern and interactive**
* **Important information was easier to find**
* **The railroad became more accessible online**

---

## User-Centered Design Lessons

The biggest lesson from this capstone was that user feedback matters more than a feature list. The team learned that not every idea should be implemented if it does not match real needs.

Examples from the project include:

* GPS tracker was removed because visitors did not need it in a small park setting.
* The AI assistant was identified as highly valuable because workers spent time answering repeated questions.
* The team used Google reviews and worker feedback to improve the website based on actual use cases.

This shows that design-based research works best when the team listens to real people and focuses on practical value.

## Summary

The PMRR project was successful because it moved the railroad website from a static information source to a more dynamic, visitor-focused digital experience. The redesign solved real issues around planning, access, community engagement, and operations. It also showed that a strong website is not just about having many features — it is about solving the right problems for the right users.
