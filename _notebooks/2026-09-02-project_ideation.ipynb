# DSA Website Problems and Improvement Plan

The current DSASD.org experience creates friction for both members and the public. Important information is spread across multiple pages, PDFs, and external links, which makes the site feel outdated and difficult to trust.

## Problems with the Current Website

The website has several issues that make it harder for visitors to find what they need:

* **Information is scattered:** Benefits, board updates, forms, and event details are not grouped in a clear way.
* **Member and public content are mixed together:** People do not know whether they are looking at information for members or for the public.
* **Search and navigation are weak:** Visitors have to click through too many pages just to find a form or document.
* **Important pages feel buried:** Donation information, community events, and member resources are not easy to find from the main experience.
* **Static PDFs are harder to use:** Many documents are posted as PDFs instead of clear, readable web pages.
* **The homepage does not guide people:** There is no clear path for what a visitor should do next.
* **The user experience feels outdated:** The site does not create a strong sense of trust, clarity, or modern organization.
* **The login button says log out when you haven't even logged in yet**
---

## How We Can Fix It

### 1. Separate the two audiences clearly
Create a visible top toggle or split navigation with two paths:

* **Public Hub:** events, donation links, community news, foundation updates
* **Member Portal:** benefits, legal resources, board docs, forms, retirement information

This makes the site easier to understand immediately.

### 2. Build a cleaner dashboard for members
Members should not have to hunt through multiple pages. We can create a single member landing page with sections for:

* Benefits and enrollment
* Legal assistance resources
* Board updates and meeting minutes
* Discounts and partner programs
* Forms and documents

This gives members one place to go for essential information.

### 3. Simplify the public experience
The public side should focus on trust and action. Include clear calls to action like:

* Donate
* View community events
* Read news updates
* Contact the foundation

This helps visitors quickly understand the purpose of the organization and how to engage.

### 4. Replace scattered PDFs with cleaner digital pages
Instead of burying information in PDFs, we can:

* Convert newsletters and updates into readable web articles
* Group related documents by category
* Add search and filter options for files
* Use simpler page layouts that are easier to scan

This improves usability and makes the site feel more modern.

### 5. Improve events and forms
The website should make it easy to find and register for events. We can add:

* A searchable event calendar
* Event categories such as meetings, charity events, and public programs
* RSVP or registration links
* Digital forms for elections, petitions, and member requests

This reduces friction and makes more actions possible online.

### 6. Improve search and homepage guidance
A strong search function and clearer homepage layout will help users instantly find information. We can add:

* Search bar in the header
* Common actions above the fold
* Clear buttons like "Member Login", "Benefits", "Donate", and "Events"
* Better breadcrumbs and page organization

### 7. Make the site feel more organized and trustworthy
The website should communicate professionalism and clarity. We can fix that by:

* Using consistent page structure
* Updating outdated content
* Creating cleaner visual hierarchy
* Highlighting key actions and resources

## Summary

The main problem is not just design—it is confusion. Visitors cannot quickly tell where to go, what is for members, and what is for the public. The fix is to organize the site around user needs, group important information, reduce reliance on hard-to-use PDFs, and give each audience a clear path.

If the site is rebuilt around clarity, searchability, and purpose, it will become far easier for members and the community to use.
