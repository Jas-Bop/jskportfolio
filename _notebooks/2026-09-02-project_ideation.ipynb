# DSA Website Ideation

This section focuses on the DSA website and the problems that make it difficult for visitors to use.

![Watch parts tracker screenshot]({{ '/images/rpg/Screenshot%20from%202026-09-03%2007-38-05.png' | relative_url }})

---

## Problems with the Current Website

The current DSASD.org experience creates friction for both members and the public. Important information is spread across multiple pages, PDFs, and external links, which makes the site feel outdated and difficult to trust.

* **Information is scattered:** Benefits, board updates, forms, and event details are not grouped in a clear way.
* **Member and public content are mixed together:** People do not know whether they are looking at information for members or for the public.
* **Search and navigation are weak:** Visitors have to click through too many pages just to find a form or document.
* **Important pages feel buried:** Donation information, community events, and member resources are not easy to find from the main experience.
* **Static PDFs are harder to use:** Many documents are posted as PDFs instead of clear, readable web pages.
* **The homepage does not guide people:** There is no clear path for what a visitor should do next.
* **The user experience feels outdated:** The site does not create a strong sense of trust, clarity, or modern organization.
* **The login button says log out when you haven't even logged in yet.**

---

## How We Can Fix It

### 1. Separate the two audiences clearly
Create a visible top toggle or split navigation with two paths:

* **Public Hub:** events, donation links, community news, foundation updates
* **Member Portal:** benefits, legal resources, board docs, forms, retirement information

### 2. Build a cleaner dashboard for members
Create a single landing page with sections for:

* Benefits and enrollment
* Legal assistance resources
* Board updates and meeting minutes
* Discounts and partner programs
* Forms and documents

### 3. Simplify the public experience
Focus on trust and action with clear calls to action like:

* Donate
* View community events
* Read news updates
* Contact the foundation

### 4. Replace scattered PDFs with cleaner digital pages
Use readable web pages instead of burying information in PDFs.

### 5. Improve events and forms
Add search, filters, categories, RSVP links, and digital forms.

### 6. Improve search and homepage guidance
Use a visible search bar and clear buttons like “Member Login,” “Benefits,” “Donate,” and “Events.”

### 7. Make the site feel more organized and trustworthy
Use consistent page structure, cleaner hierarchy, and stronger visual contrast.

```mermaid
flowchart TD
    A[Visitor lands on homepage] --> B{What are they looking for?}
    B -->|Member resources| C[Member Portal]
    B -->|Public information| D[Public Hub]
    C --> E[Benefits, forms, legal docs]
    D --> F[Events, news, donations]
    E --> G[Action complete]
    F --> G
    G --> H[Trust restored]
```

## Summary

The main problem is confusion. Visitors cannot quickly tell where to go, what is for members, and what is for the public. The fix is to organize the site around user needs, group important information, reduce reliance on PDFs, and give each audience a clear path.

# DED Website Ideation

This section focuses on the Doing Exceptional Deeds website and the ways it could be improved to create a clearer and more effective experience.

---

## Problems with the Current Website

The Doing Exceptional Deeds website has a strong mission and a clear purpose, but the overall experience is crowded and difficult to navigate. The site communicates a lot of important information, but it does not always guide users toward the most important next step.

* **Too much information on the homepage:** The mission, impact stats, program list, donations, stories, and calls to action all compete for attention.
* **The main message is not prioritized:** The homepage has a strong message, but the layout does not consistently lead users to a clear action path.
* **Programs and organization information are blended together:** It is hard to tell which content is for parents, students, volunteers, donors, and partners.
* **Important actions are buried:** Donation, volunteer, partner, and contact options appear in multiple places without a consistent user flow.
* **The site feels visually crowded:** Many blocks, links, images, and sections overlap, making the experience feel busy instead of focused.
* **The hierarchy is unclear:** Some pages feel like marketing pages while others feel like directories or forms, creating a less consistent brand experience.
* **There are many entry points but not a clear journey:** A visitor may not know whether they should donate, register, get involved, contact the organization, or explore programs first.

```mermaid
mindmap
  root((D.A.D. Website Problems))
    Message
      unclear priorities
      too many competing headlines
      weak hero focus
    Navigation
      audiences mixed together
      unclear next steps
      too many links
    Content
      crowded homepage
      repeated calls to action
      inconsistent hierarchy
    Action
      donation confusion
      volunteer confusion
      partner confusion
```

---

## How We Can Fix It

### 1. Define the homepage goal more clearly
The homepage should answer one question immediately: “What is this organization, and what should I do next?”

A better structure could be:

* Who we are
* What we do
* How to get involved
* Ways to support us

### 2. Separate audiences by user need
Create distinct pathways for:

* Parents and students
* Volunteers
* Donors
* Partners and sponsors

### 3. Simplify the navigation
Use a cleaner navigation menu with categories like:

* Programs
* About
* Impact
* Get Involved
* Donate
* Contact

### 4. Reduce homepage clutter
Keep the homepage focused on:

* mission statement
* featured programs
* impact numbers
* three simple action buttons
* real stories
* final call to action

### 5. Make the action flow obvious
Add visible CTA blocks for:

* Donate Now
* Explore Programs
* Volunteer
* Become a Partner
* Contact Us

### 6. Create stronger program landing pages
Each program page should include:

* purpose
* audience
* how it works
* impact
* how to join or support

### 7. Improve visual hierarchy and trust
Use stronger spacing, cleaner cards, clear headings, and fewer overlapping elements to make the site feel more professional.

```mermaid
flowchart TD
    A[Visitor arrives] --> B{What are they here for?}
    B -->|Programs| C[Programs page]
    B -->|Volunteer| D[Volunteer page]
    B -->|Donate| E[Donation page]
    B -->|Partner| F[Partner page]
    C --> G[Understand program and next step]
    D --> G
    E --> G
    F --> G
    G --> H[Clear action taken]
```

## Summary

Doing Exceptional Deeds has a powerful mission, but the website currently feels overcrowded and harder to navigate than it needs to be. The biggest issue is not the mission—it is the lack of a clear user journey. If the site simplifies its homepage, separates audiences, strengthens calls to action, and creates cleaner program pages, it will be much easier for families, volunteers, donors, and partners to understand and support the organization.
